# **BOE and Istitlaa resorces cleaning and preprocessing**

Here a comprehensive data cleaning process was applied to the Saudi Labor Law datasets collected from two different sources:

boe_laborlaw_articles_Edit.csv from the BOE website.

istitlaa_laborlaw.csv from the Istitlaa platform.

The cleaning was implemented using Python, with the pandas library for data manipulation and the re (regular expressions) library for text normalization.
The main steps performed were as follows:

Loading the datasets using the utf-8-sig encoding to avoid BOM-related issues.

Creating copies of the original data to preserve the raw versions.

Removing duplicates to ensure that each record is unique.

Replacing missing values with "N/A" to prevent processing errors.

Applying the clean_text function to standardize and clean textual data by:

Converting all text to lowercase for consistency.

Removing colons (:) and unnecessary characters.

Deleting newlines and tabs (\n, \r, \t).

Replacing multiple spaces with a single space for cleaner formatting.

Saving the cleaned datasets into two new CSV files:

boe_cleaned.csv

istitlaa_cleaned.csv

As a result, both datasets are now cleaned, standardized, and ready for further analysis or merging in the next stages of the project.

```python
import pandas as pd
import re
# Load both datasets with encoding handling
boe = pd.read_csv("boe_laborlaw_articles_Edit.csv", encoding="utf-8-sig")
istitlaa = pd.read_csv("istitlaa_laborlaw.csv", encoding="utf-8-sig")

# Function to clean text
def clean_text(text):
    if isinstance(text, str):
        text = text.lower().strip()
        text = text.replace(":", "")  # يحذف النقطتين
        text = re.sub(r'[\r\n\t]+', ' ', text)  # يحذف الأسطر الجديدة والتبويبات
        text = re.sub(r'\s{2,}', ' ', text)  # يحول المسافات الزايدة لمسافة وحدة
        return text
    return text


# Cleaning BOE dataset
boe_cleaned = boe.copy()
boe_cleaned = boe_cleaned.drop_duplicates()
boe_cleaned = boe_cleaned.fillna("N/A")
for col in boe_cleaned.select_dtypes(include='object').columns:
    boe_cleaned[col] = boe_cleaned[col].apply(clean_text)

# Cleaning Istitlaa dataset
istitlaa_cleaned = istitlaa.copy()
istitlaa_cleaned = istitlaa_cleaned.drop_duplicates()
istitlaa_cleaned = istitlaa_cleaned.fillna("N/A")
for col in istitlaa_cleaned.select_dtypes(include='object').columns:
    istitlaa_cleaned[col] = istitlaa_cleaned[col].apply(clean_text)



# Save cleaned datasets
boe_cleaned_path = "boe_cleaned.csv"
istitlaa_cleaned_path = "istitlaa_cleaned.csv"

boe_cleaned.to_csv(boe_cleaned_path, index=False, encoding="utf-8-sig")
istitlaa_cleaned.to_csv(istitlaa_cleaned_path, index=False, encoding="utf-8-sig")

boe_cleaned_path, istitlaa_cleaned_path
```
