```python
# ===== 1) Imports + Load Data =====
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# تحميل ملفات البنت 1 (Baseline)
train_X = pd.read_csv("x_train.csv")   # فيه عمود text
train_y = pd.read_csv("y_train.csv")   # فيه عمود source
test_X  = pd.read_csv("x_test.csv")
test_y  = pd.read_csv("y_test.csv")

# نحولهم لـ Series واضحة
x_train = train_X["text"]
y_train = train_y["source"]
x_test  = test_X["text"]
y_test  = test_y["source"]

print(f"Train samples: {len(x_train)}, Test samples: {len(x_test)}")
```

TF-IDF

```python
# ===== 2) TF-IDF Encoding =====
tfidf = TfidfVectorizer(
    max_features=6000,
    ngram_range=(1, 2)
)

X_train_tfidf = tfidf.fit_transform(x_train)
X_test_tfidf  = tfidf.transform(x_test)

print("TF-IDF encoding done!")
```

Model 1 (Random Forest)

```python
# ===== 3) Build & Train Model 1 (Random Forest) =====
model1 = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    random_state=42
)

model1.fit(X_train_tfidf, y_train)
print("Model 1 (Random Forest) has been trained!")
```

```python
# ===== 4) Evaluation & Saving Results =====
# التنبؤ على بيانات الاختبار
y_pred_model1 = model1.predict(X_test_tfidf)

# حساب الـ Accuracy
accuracy = accuracy_score(y_test, y_pred_model1)
print(f"Model 1 Accuracy: {accuracy:.4f}")

# تقرير تصنيفي مفصل
report = classification_report(
    y_test,
    y_pred_model1,
    output_dict=True,
    zero_division=0
)
report_df = pd.DataFrame(report).transpose()
print("\nClassification report (Model 1):")
print(report_df)

# حفظ النتائج في ملف CSV ليستفيدون منها البنات 4 و 5
report_df.to_csv("model1_results.csv", encoding="utf-8-sig")
print("\nModel 1 results saved as 'model1_results.csv'")
```

Model 2 (SVM algorithm)




```python
# ===== 1) Import SVM =====
from sklearn.svm import SVC

# ===== 2) Build & Train Model 2 (SVM) =====
model2 = SVC(kernel="linear", probability=False)  # Linear kernel best for TF-IDF
model2.fit(X_train_tfidf, y_train)
print("Model 2 (SVM) has been trained!")
```

```python
# ===== 3) Evaluation =====
y_pred_model2 = model2.predict(X_test_tfidf)
# Compute accuracy
accuracy2 = accuracy_score(y_test, y_pred_model2)
print(f"Model 2 Accuracy: {accuracy2:.4f}")
# Generate detailed classification report for each class (precision, recall, f1-score)
report2 = classification_report(y_test, y_pred_model2, output_dict=True, zero_division=0)
report_df2 = pd.DataFrame(report2).transpose()
print("\nClassification Report (Model 2 - SVM):")
print(report_df2)

# Save results for group use
report_df2.to_csv("model2_results.csv", encoding="utf-8-sig")
print("Model 2 results saved as 'model2_results.csv'")
```
